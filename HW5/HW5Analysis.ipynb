1. From a medical standpoint, certain values cannot be 0 so these 0's in the dataset are placeholder values. In other features, such as pregnancies 0 is an acceptable answer and cannot be viewed as a placeholder. Consequently, I used mediam imputation to ignore any placeholders which might cause an error or inaccuracy later when building different models. If instead I deleted the rows, relevant data could be lost since the entire row of information is removed to counteract potentially one column's 0.

2. Hard voting focuses on what the majority points to while soft voting takes an average. Calibrating is necessary to run a soft vote since it relies on probabilities that are scaled to eachother. When there is no calibration, soft-vote becomes less accurate than hard voting in most cases.

3. Out of Bag evaluation allows each training set to read trees that were not accounted for initially, and make a prediction off of it. This is then compared to the in bag data and cross evaluated. It is efficient since it takes the trees that were missed instead of adding more training cycles.

4. Bagging needs random rows since trees are initially correlated. The random forests creates better generalizations since the feature importances balance out in each split.

5. AdaBoost often is overfit than bagging since it builds one stump at a time. Since it occurs in order, it is prone to carry on errors after time. It starts of underfitting, and when the stumps happen for too long it begins to be overfitting. You can slow the learning rate to see when the start of that error begins.
