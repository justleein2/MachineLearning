**Contrast the operational mechanisms of standard DBSCAN and HDBSCAN. How does HDBSCAN's hierarchical, stability-based approach overcome DBSCAN's vulnerability to a single global eps parameter when clustering chemical datasets with varying local densities?**

DBSCAN checks if a point is in a cluster by checking a minimum number of nearby points within the eps. Since all the clusters are judged by the same eps, it is hard to check for dense or sparse clusters at the same time. You need small eps to find dense clusters, but the sparse clusters will become noise.

HDBSCAN doesn't use eps and instead uses density level hierarchy to measure local density to its closest neighbour. Using a dendrogram can show how these clusters form based on each branch. The wider the range of density levels, the more stable the branch is. Since the branches are relative to your data instead of global values, it can evaluate both dense and sparse clusters. In chemical data this is important because samples naturally have a lot of variation so setting a global standard ignores significant density changes.
