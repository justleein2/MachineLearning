2. The dataset hardcodes the missing values as -200 which can alter your final data and the validaty of your models. To counteract this, you have to replace all values of -200 to NaN, and clear space for the actual data to not be misrepresented. Additionally, using median as a strategy will ignore the large outliers that would affect the mean.

3. Pipeline and ColumnTransformer can prevent data leakage by focusing on a trained set of data. It also helps separate the numerical and categorical focuses for training and targeted manipulation.

4. StandardScaler and MinMaxScaler both rescale the features to a different setting. StandardScaler allows the distribution of data to be preserved and MinMaxScaler will set the given bounds of all features in the target dataset. These can be important because they influence the accuracy of the final metrics. StandardScaler ended up needing less iterations for logistic regression. MinMaxScaler also influenced the coefficients needed for Ridge.

5. SDGClassifier uses small batches of data to approximate the data through multiple iterations. While closed-form will make the approximation based on the entire set of data in one go. This means that SDG is more flexible over time and can handle things such as hinge and log loss, which logistic regression cannot do. Accuracy and convergences can be used to determine which model is better for the given set of data.

6. Since Ridge shrinks the model towards 0 but never reaches it, it is important for checking the strength of influence different features have. Lasso can become 0 so it can remove features that are weak. Depending on the operational decision, you would want to look at one or the other, but the larger you raise the alpha the less accurate the data will become as seen when Ridge was alpha 10 and Lasso was alpha 1.
